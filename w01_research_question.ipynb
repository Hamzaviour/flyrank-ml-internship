## 2. The question: decision, action, cost of a wrong call

**Research question**

Can observable search, content, and engagement signals be used to rank which content pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?

**Unit of analysis**

One pseudonymized content item/page.

**Decision**

Which pages should a limited content/SEO team review first?

**Action**

Use the ranked output to prioritize human review and suggest an evidence-based action such as refresh, expansion, CTR review, protection, pruning/merge review, or monitoring.

**Cost of a wrong recommendation**

A false positive may waste limited content resources on a page with little opportunity, while a false negative may cause a potentially valuable page to be overlooked. Because review capacity is limited, performance at the top of the ranking is particularly important.

**Why ML?**

Multiple observable signals may interact in ways that are difficult to capture with a small set of fixed rules. A learned model may identify combinations of signals that improve ranking quality over a transparent baseline. However, this must be tested through leakage-safe validation rather than assumed.
